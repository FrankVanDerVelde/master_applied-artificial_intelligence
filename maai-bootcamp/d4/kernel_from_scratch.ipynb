# Kernel operation from scratch

In this notebook, we will implement a simple image processing operation using a convolutional kernel from scratch. 

Tested with python 3.12.

We will use the PIL image library to load and display images, and Numpy for numerical operations.


```python
from PIL import Image
import numpy as np
```

```python
# Load a color image change this to your image path
img = Image.open("./images/test_image.png")

# Convert to grayscale
img_gray = img.convert('L')   # 'L' mode = (8-bit pixels, grayscale)

# If you want a numpy array of gray values:
gray_array = np.array(img_gray)  # shape [H, W], dtype=uint8

# Show the gray image
img_gray.show()
```

```python
# STUDY THE FUNCTION!

def kernel_operation(img_gray, Kernel):
    h, w = img_gray.shape
    p = np.pad(img_gray, 1, mode="edge")
    out = sum(Kernel[i, j] * p[i:i+h, j:j+w] for i in range(3) for j in range(3))
    out = (out - out.min()) / (np.ptp(out) + 1e-8) * 255
    return out.astype(np.uint8)
```

```python
# === 3x3 kernel for vertical edge detection ===
Kernel = np.array(  [[-1, 0, 1],
                    [-1, 0, 1],
                    [-1, 0, 1]], dtype=np.float32)  # verticale randen


# Call the function:
result = kernel_operation(gray_array, Kernel)
new_image = Image.fromarray(result)
new_image.show()

# save
new_image.save("./images/edge_detection.png")
```

```python
# Exercise 1
# Horizontal edge detection kernel
Kernel_horizontal = np.array([[1,  1,  1],
                               [0,  0,  0],
                               [-1, -1, -1]], dtype=np.float32)

# Call the function:
result_horizontal = kernel_operation(gray_array, Kernel_horizontal)
new_image_horizontal = Image.fromarray(result_horizontal)
new_image_horizontal.show()

# Save the result
new_image_horizontal.save("./images/horizontal_edge_detection.png")
```

```python
# Exercise 2
# Sobel filter for edge detection

# Sobel kernel for detecting vertical edges
Kernel_sobel_vertical = np.array([[-1, 0, 1],
                                   [-2, 0, 2],
                                   [-1, 0, 1]], dtype=np.float32)

# Sobel kernel for detecting horizontal edges
Kernel_sobel_horizontal = np.array([[ 1,  2,  1],
                                     [ 0,  0,  0],
                                     [-1, -2, -1]], dtype=np.float32)

# Apply the vertical Sobel kernel
result_sobel_vertical = kernel_operation(gray_array, Kernel_sobel_vertical)
new_image_sobel_vertical = Image.fromarray(result_sobel_vertical)
new_image_sobel_vertical.show()

# Save the vertical Sobel result
new_image_sobel_vertical.save("./images/sobel_vertical_edge_detection.png")

# Apply the horizontal Sobel kernel
result_sobel_horizontal = kernel_operation(gray_array, Kernel_sobel_horizontal)
new_image_sobel_horizontal = Image.fromarray(result_sobel_horizontal)
new_image_sobel_horizontal.show()

# Save the horizontal Sobel result
new_image_sobel_horizontal.save("./images/sobel_horizontal_edge_detection.png")
```

```python
# Exercise 3
# 5x5 Gaussian blur kernel
Kernel_gaussian = np.array([[1,  4,  6,  4, 1],
                             [4, 16, 24, 16, 4],
                             [6, 24, 36, 24, 6],
                             [4, 16, 24, 16, 4],
                             [1,  4,  6,  4, 1]], dtype=np.float32)

# Normalize the kernel
Kernel_gaussian /= Kernel_gaussian.sum()

# Apply the Gaussian blur kernel
result_gaussian = kernel_operation(gray_array, Kernel_gaussian)
new_image_gaussian = Image.fromarray(result_gaussian)
new_image_gaussian.show()

# Save the Gaussian blur result
new_image_gaussian.save("./images/gaussian_blur.png")
```

```python
# Bonus exercise
# Unsharp Masking kernel
Kernel_unsharp = np.array([[  1,   4,    6,   4,   1],
                            [  4,  16,   24,  16,   4],
                            [  6,  24, -476,  24,   6],
                            [  4,  16,   24,  16,   4],
                            [  1,   4,    6,   4,   1]], dtype=np.float32)

# Normalize the kernel
Kernel_unsharp /= -256

# Apply the Unsharp Masking kernel
result_unsharp = kernel_operation(gray_array, Kernel_unsharp)
new_image_unsharp = Image.fromarray(result_unsharp)
new_image_unsharp.show()

# Save the Unsharp Masking result
new_image_unsharp.save("./images/unsharp_masking.png")
```

```python
# Crazy vibrant color effect
# Load the original color image
img_color = np.array(img)  # Convert the image to a NumPy array

# Define different kernels for each channel
Kernel_sharpen = np.array([[ 0, -1,  0],
                           [-1,  5, -1],
                           [ 0, -1,  0]], dtype=np.float32)

Kernel_edge = np.array([[-1, -1, -1],
                        [-1,  8, -1],
                        [-1, -1, -1]], dtype=np.float32)

Kernel_blur = np.array([[1,  4,  6,  4, 1],
                        [4, 16, 24, 16, 4],
                        [6, 24, 36, 24, 6],
                        [4, 16, 24, 16, 4],
                        [1,  4,  6,  4, 1]], dtype=np.float32)
Kernel_blur /= Kernel_blur.sum()  # Normalize the blur kernel

# Function to apply different kernels to each channel
def apply_different_kernels(img_color, kernels):
    R, G, B = img_color[:, :, 0], img_color[:, :, 1], img_color[:, :, 2]
    
    # Apply different kernels to each channel
    R_result = kernel_operation(R, kernels[0])
    G_result = kernel_operation(G, kernels[1])
    B_result = kernel_operation(B, kernels[2])
    
    # Stack the channels back together
    return np.stack([R_result, G_result, B_result], axis=-1)

# Apply the kernels to the color image
result_crazy_colors = apply_different_kernels(img_color, [Kernel_sharpen, Kernel_edge, Kernel_blur])

# Convert the result back to an image and display it
new_image_crazy_colors = Image.fromarray(result_crazy_colors)
new_image_crazy_colors.show()

# Save the result
new_image_crazy_colors.save("./images/crazy_vibrant_colors.png")
```
