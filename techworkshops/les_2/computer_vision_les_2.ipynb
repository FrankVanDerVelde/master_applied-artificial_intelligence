```python
# Use a pipeline as a high-level helper
from transformers import pipeline

pipe = pipeline("image-classification", model="microsoft/resnet-50")
pipe("https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/hub/parrots.png")
```

```python
from datasets import load_dataset

# Login using e.g. `huggingface-cli login` to access this dataset
ds = load_dataset("MichielBontenbal/elephants")
```

```python
result = pipe(ds['train']['image'][0])
result[0]['label'], result[0]['score']
```

```python
import random
import matplotlib.pyplot as plt

# Select 10 random indices from the training set
indices = random.sample(range(len(ds['train'])), 10)

fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for ax, idx in zip(axes.flatten(), indices):
    image = ds['train']['image'][idx]
    label = ds['train']['label'][idx]
    ax.imshow(image)
    ax.set_title(f"Label: {label}")
    ax.axis('off')
plt.tight_layout()
plt.show()
```
